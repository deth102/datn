```python
import numpy as np
import pandas as pd #đọc đầu vào csv
import matplotlib.pyplot as plt #vẽ đồ thị
from scipy.signal import spectrogram # biểu diễn tren miền thời gian tần số
```

 Rotating speed of the motor in the normal state

```python
# Đọc dữ liệu tốc độ
speed_data = pd.read_csv("rpm_normal_0.csv")

# Vẽ đồ thị
plt.figure(figsize=(12,3))
plt.plot(speed_data["time"]/60, speed_data["rpm"], linewidth=1)
plt.xlabel("Time (min)")
plt.ylabel("Rotating speed (RPM)")
plt.title("Rotating Speed vs Time in the normal state")
plt.show()
```

Timeseries acceleration data with constant rotating speed in normal state

```python
fs = 25600  # Tần số lấy mẫu

# Đọc dữ liệu
vibration_constant_normal = pd.read_csv("vibration_normal_constant.csv")

dataplot_constant_normal= vibration_constant_normal.iloc[:, 0].values

Time = np.arange(len(dataplot_constant_normal)) / fs
# Vẽ dữ liệu dao động
plt.figure(figsize=(12,3))
plt.plot(Time/60, dataplot_constant_normal, color='black', linewidth=0.8)
plt.xlabel("Time (min)")
plt.ylabel("Amplitude (g)")
plt.title("Acceleration data with constant rotating speed in normal state")
plt.show()
```

Corresponding spectrogram with the constant rotating speed in Normal state

```python
f, t, Sxx = spectrogram(dataplot_constant_normal, fs)
plt.figure(figsize=(12, 3))
plt.pcolormesh(t/60, f/1000, 10*np.log10(Sxx), shading='gouraud')
plt.ylabel("Frequency (kHz)")
plt.xlabel("Time (min)")
plt.title("Time-Frequency domain with the constant rotating speed in Normal state")
plt.colorbar(label="Power [dB]")
plt.show()
```
