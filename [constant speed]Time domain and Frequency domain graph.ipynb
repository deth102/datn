Thư viện

```python
import numpy as np
import pandas as pd #đọc đầu vào csv
import matplotlib.pyplot as plt #vẽ đồ thị
from scipy.signal import spectrogram # biểu diễn tren miền thời gian tần số
```

```python
fs = 25600  # Tần số lấy mẫu
```

Time domain vibration under constant at normal state

```python
normalstate= pd.read_csv("Vibration data/4Nm_Normal.csv") 
signal_normalstate = normalstate["Xa"].values
Time = np.arange(len(signal_normalstate)) / fs
# Vẽ dữ liệu dao động
plt.figure(figsize=(12,3))
plt.plot(Time/60, signal_normalstate, color='black', linewidth=0.8)
plt.xlabel("Time (min)")
plt.ylabel("Amplitude (g)")
plt.title("Normal state")
plt.show()
```

The segment of the normal state signal

```python
t_start = 0.0      # thời điểm bắt đầu 
t_end   = 0.08     # thời điểm kết thúc 
start = int(fs * t_start)
end   = int(fs * t_end)

Time_segment = Time[start:end]
segment = signal_normalstate[start:end]

plt.figure(figsize=(12,3))
plt.plot(Time_segment, segment, color='black', linewidth=0.8)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (g)")
plt.title("The segment of the signal - Normal state")
plt.xlim(t_start, t_end)
plt.tight_layout()
plt.show()
```

Frequency domain vibration under constant speed at normal state

```python
N = len(signal_normalstate)
fft_vals = np.fft.fft(signal_normalstate - np.mean(signal_normalstate))
fft_freqs = np.fft.fftfreq(N, 1/fs)
idx = fft_freqs > 0
plt.figure(figsize=(13, 4))
plt.plot(fft_freqs[idx], np.abs(fft_vals[idx]) * 2 / N, color='red', linewidth=0.8)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.title("Frequency Spectrum - Normal State")
plt.xlim(0, 10000)
plt.ylim(0, 0.8)
plt.grid(True)
plt.tight_layout()
plt.show()
```
